### **Primer ejercicio**

#### **Síntesis mediante Foster serie**

Para hallar la topología utilizamos la expresión del desarrollo de Foster:

$$Z(s) = k_{\infty} s + \frac{k_0}{s} + \sum \frac{2k_i s}{s^2 + w^2_i}$$

Donde:

* $k_{\infty}$ representa el valor de un inductor en serie.
* $k_0$ representa el valor de un capacitor en serie.
* $\sum \frac{2k_i s}{s^2 + w^2_i}$ representa un tanque ***LC*** en serie.

La impedancia $F(s)$ factorizada queda:

$$F(s) = \frac{ (s^2 + 1) (s^2 + 3) }{ s(s^2+2) } $$

Ahora procedemos a calcular los residuos de nuestra impedancia para hallar los valores de los componentes:

**Residuo en cero ($k_0$)**

$$k_0 = \lim_{s -> 0} sZ(s)$$

$$k_0 = \lim_{s -> 0} s \frac{ (s^2 + 1) (s^2 + 3) }{ s(s^2+2) } $$

$$k_0 = \lim_{s -> 0} \frac{ (s^2 + 1) (s^2 + 3) }{ (s^2+2) } $$

$$k_0 = \frac{3}{2}$$

* $k_0$ es un capacitor $C=\frac{2}{3}$


**Residuo en infinito ($k_\infty$)**

$$k_\infty = \lim_{s -> \infty} \frac{Z(s)}{s}$$

$$k_\infty = \lim_{s -> 0} s \frac{ (s^2 + 1) (s^2 + 3) }{ s^2(s^2+2) } $$

$$k_\infty = 1$$

* $k_\infty$ es un inductor $L=1$

**Residuo en los polos**

$$2k_i = \lim_{s^2->-2} \frac{s^2 + 2}{s} Z(s)$$

$$2k_i = \lim_{s^2->-2} \frac{s^2 + 2}{s} \frac{ (s^2 + 1) (s^2 + 3) }{ s(s^2+2) } $$

$$2k_i = \lim_{s^2->-2} \frac{ (s^2 + 1) (s^2 + 3) }{ s^2 } $$

$$2k_i = \frac{1}{2}$$

Con este valor igualamos la expresion de Foster con la impedancia de un tanque **LC**

$$\frac{ s\frac{1}{C} }{ s^2 + \frac{1}{LC} } = \frac{2k_i s}{s^2 + w^2_i}$$

Entonces:

* $C = 2$
* $L = \frac{1}{4}$


**Circuito Final**

Agrupando los elementos nos queda armado el circuito final:
* $C_0 = \frac{2}{3}$
* $L_{\infty} = 1$
* Tanque **LC**:
  * $C_1 = 2$
  * $L_2 = \frac{1}{4}$

$$Y(s) = \frac{ s(s^2 + 2) }{ (s^2+1)(s^2+3) }$$

IMAGEN DE LOS POLOS Y CEROS

$$2k_1 = \lim_{s^2->-1} \frac{ s^2 + 2 }{ s^2 + 1} $$

#### **Síntesis mediante Cauer remoción en infinito**

Para aplicar la remoción de Cauer en el infinot debemos dividir los polinomios de la función inmitancia.

Si consideramos que $F(s)=Z(s)$ el primer elemento removido será un inductor en serie, mientras que el segundo será en derivación. Esto ocurre debido a que al remover un elemento de la función, pasamos de $Z(s)$ a $Y(s)$.

Si $F(s)=\frac{ s^4 + 4s^2 + 3 }{ s(s^2 +2) }$

Extraemos el primer elemento:

$$
\begin{array}{r|l}
s^4 + 4s^2 + 3 & s^3 + 2s \\
\hline & \mathbf{s} \\
-(s^4 + 2s^2) \phantom{+ 3} \\
\hline
\mathbf{2s^2 + 3} & 
\end{array}
$$

El cociente $s$ indica un inductor serie $L_1=1$

Ahora dividimos el denominador anterior por el nuevo resto (estamos con $Y(s)$):

$$
\begin{array}{r|l}
s^3 + 2s & 2s^2 + 3 \\
\hline & \mathbf{\frac{1}{2}s} \\
-(s^3 +  \frac{3}{2}s) \phantom{0} \\
\hline
\mathbf{\frac{1}{2}s} & 
\end{array}
$$

El cociente $\frac{1}{2}s$ indica un capacitor derivación $C_1=\frac{1}{2}$

Ahora dividimos el denominador anterior por el nuevo resto (volvemos a $Z(s)$):

$$
\begin{array}{r|l}
2s^2 + 3 & \frac{1}{2}s \\
\hline & \mathbf{4s} \\
-2s^2    \phantom{+3} \\
\hline
\mathbf{3} & 
\end{array}
$$

El cociente $4s$ indica un inductor serie $L_2=4$

Ahora dividimos el denominador anterior por el nuevo resto (volvemos a $Y(s)$):

$$
\begin{array}{r|l}
\frac{1}{2}s & 3 \\
\hline & \mathbf{\frac{1}{6}s} \\
-\frac{1}{2}s    \phantom{0} \\
\hline
\mathbf{0} & 
\end{array}
$$

El cociente $\frac{1}{6}s$ indica un capacitor derivación $C_2=\frac{1}{6}$

**Circuito Final**

Agrupando los elementos nos queda armado el circuito final:

* $L_1 = 1$ serie.
* $C_1 = \frac{1}{2}$ derivación.
* $L_2 = 4$ serie.
* $C_2 = \frac{1}{6}$ derivación.

#### **Síntesis mediante Cauer remoción en DC**

Para aplicar la remoción de Cauer en 0 uso su segunda versión (Cauer 2), la cual me permite eliminar las singularidades en el origen ordenando los coeficientes del numerador y el denominador de la función de menor a mayor:

Ahora la función inmitancia $F(s)$ queda como:

$$F(s) = \frac{ 3 + 4s^2 + s^4 }{ 2s + 3s^3 }$$

Con esta función procedemos de igual manera que con Cauer 1:

Extraemos el primer elemento:

$$
\begin{array}{r|l}
3 + 4s^2 + s^4 &  2s + s^3 \\
\hline & \mathbf{\frac{3}{2s}} \\
-(3 + \frac{3}{2}s^2) \phantom{+ 3} \\
\hline
\mathbf{\frac{5}{2}s^2 + s^4} & 
\end{array}
$$

El cociente $\frac{3}{2s}$ indica un capacitor serie $C_1=\frac{2}{3}$

Ahora dividimos el denominador anterior por el nuevo resto (estamos con $Y(s)$):

$$
\begin{array}{r|l}
2s + s^3 &  \frac{5}{2}s^2 + s^4 \\
\hline & \mathbf{\frac{4}{5s}} \\
-(2s + \frac{4}{5}s^3) \phantom{0} \\
\hline
\mathbf{\frac{1}{5}s^3} & 
\end{array}
$$

El cociente $\frac{4}{5s}$ indica un inductor derivación $L_1=\frac{5}{4}$

Ahora dividimos el denominador anterior por el nuevo resto (volvemos a $Z(s)$):

$$
\begin{array}{r|l}
\frac{5}{2}s^2 + s^4 &  \frac{1}{5}s^3 \\
\hline & \mathbf{\frac{25}{2s}} \\
-\frac{5}{2}s^2  \phantom{+3} \\
\hline
\mathbf{s^4} & 
\end{array}
$$

El cociente $\frac{25}{2s}$ indica un capacitor serie $C_2=\frac{2}{25}$

Ahora dividimos el denominador anterior por el nuevo resto (volvemos a $Y(s)$):

$$
\begin{array}{r|l}
\frac{1}{5}s^3 &  s^4 \\
\hline & \mathbf{\frac{1}{5s}} \\
-\frac{1}{5}s^3  \phantom{0} \\
\hline
\mathbf{0} & 
\end{array}
$$

El cociente $\frac{1}{5s}$ indica un capacitor serie $L_2=5$

**Circuito Final**

Agrupando los elementos nos queda armado el circuito final:

* $C_1 = \frac{2}{3}$ serie.
* $L_1 = \frac{5}{4}$ derivación.
* $C_2 = \frac{2}{25}$ serie.
* $L_2 = 5$ derivación.

### **Tercer ejercicio**

En este ejercicio tenemos a $R_1$ en serie con dos tanques: uno RL y otro RC.

La impedancia propuesta es:

$$Z(s) = \frac{ s^2 +6s + 8 }{ s^2 + 6s + 5 } = \frac{ (s+2)(s+4) }{ (s+1)(s+5) }$$

Si la analizamos en $s=0$ y $s->\infty$:

$$Z(0) = \frac{ 8 }{ 5 } = R_1 + R_3$$
$$Z(\infty) = 1 = R_1 + R_2$$

Viendo la recta de polos y ceros notamos que los tanques tienen polos simples:

IMAGEND DE LA RECTA

Para obtener los componentes podemos utilizar Foster sin los componentes $k_0$ y $k_{\infty}$ ya que solo hay tanques:

$$Z(s) = \sum \frac{k_i s}{s + w_i}$$

En nuestra red:

$$Z(s) = \frac{ R_2 s}{s + \frac{ R_2 }{ L } }+ \frac{ \frac{1}{C} }{s + \frac{ 1 }{ R_3 C }}$$

Procedemos a obtener los residuos:

* Para $s=-1$

$$k_1 = \lim_{s->-1} (s + 1) Z(s)$$

$$k_1 = \lim_{s->-1} (s + 1) \frac{ (s+2)(s+4) }{ (s+1)(s+5) }$$

$$k_1 = \lim_{s->-1} \frac{ (s+2)(s+4) }{ s+5 }$$

$$k_1 = \frac{3}{4} $$

* Para $s=-5$

$$k_2 = \lim_{s->-5} (s + 5) Z(s)$$

$$k_2 = \lim_{s->-5} (s + 5) \frac{ (s+2)(s+4) }{ (s+1)(s+5) }$$

$$k_2 = \lim_{s->-5} \frac{ (s+2)(s+4) }{ s+1 }$$

$$k_2 = -\frac{3}{4} $$

Como $K_2$ es negativo, no puede ser ningun componente por lo que expandimos $\frac{Z(s)}{s}$

$$k'_2 = \lim_{s->-5} \frac{(s + 5)}{s} \frac{ (s+2)(s+4) }{ (s+1)(s+5) }$$
$$k'_2 = \lim_{s->-5} \frac{1}{s} \frac{ (s+2)(s+4) }{ s+1 }$$
$$k'_2 = \frac{3}{20}$$


Igualando las expresionesde Foster con la de cada tanque obtenemos los componentes:

$$\frac{ \frac{1}{C} }{s + \frac{ 1 }{ R_3 C }} = \frac{k_1}{s+1}$$

$$ \frac{ R_2 s}{s + \frac{ R_2 }{ L } } = \frac{k_2s}{s+5}$$

Asociamos al polo $s=-1$ al tanque RC dado que es la singularidad mas cercana al origen.

Donde:
* $R_2=\frac{3}{20}$
* $R_3=\frac{3}{4}$
* $L_1=\frac{3}{100}$
* $C_1=\frac{4}{3}$

Siendo $R_1 = Z(\infty) - R_2 = 1 - R_2$
